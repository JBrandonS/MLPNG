```python
run_name = "new+bias"
core = Core(["settings/n128.json", "--nsims", "10000"], log_level=0)
ds = MapDataset.fromCore(core)
train, val, test = ds.split(
    0.4,
    0.1,
    0.5,
    to_tf=True,
    batch_size=64,
    duplicates=[25, 10, 2],
    gen_batch_size=16,
    cache_file=f"{core.name}-043025-{run_name}-{core.shapes_str()}",
)

# the paper does not rotate the test set
test.rotate = False

likelihoods = core.get_likelihoods()
```

```python
model_file = "/users/stevensonb/Research/MLPNG/data/models/l383_n128_T_10000-new+bias-298686.keras"
new_model = tf.keras.models.load_model(model_file)
y_test = np.concatenate([y for _, y in test])
preds = new_model.predict(test, verbose=0)
metrics = new_model.evaluate(test, verbose=0)

make_trainer_plots(
    core, "", run_name, None, metrics, y_test, preds, likelihoods=likelihoods
)
```
